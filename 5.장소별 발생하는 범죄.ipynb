```python
df_delete = df.drop('전체범죄')
```

```python
values2 = df_delete.groupby('범죄대분류')['단독주택'].sum().values
values2
```

```python
labels2 = df_delete.groupby('범죄대분류')['단독주택'].sum().index
labels2
```

```python
type(labels2)
```

```python
labels = labels2
values = values2

total1 = sum(values)

new_labels = []
new_values = []

etc = 0

for i in range(len(values)):
    if values[i]/total1*100 >= 2:
        new_labels.append(labels[i])
        new_values.append(values[i])
    else:
        etc+=values[i]

new_labels.append('기타')
new_values.append(etc)
```

```python
wedgeprops = {'width':0.6}
explode = [0.1]*len(new_values)
colors = ['dimgrey','grey','darkgrey','silver','lightgray','whitesmoke','white']
plt.pie(new_values,labels = new_labels, autopct = '%.1f%%',colors = colors,textprops = {'fontsize':8},startangle = 90,explode = explode,wedgeprops = wedgeprops)
plt.title('단독주택에서 일어나는 범죄',color = 'red', fontsize = 7)
plt.show()
```

```python
values3 = df_delete.groupby('범죄대분류')['아파트'].sum().values
labels3 = df_delete.groupby('범죄대분류')['아파트'].sum().index

labels = labels3
values = values3

total2 = sum(values)

new_labels2 = []
new_values2 = []

etc = 0

for i in range(len(values)):
    if values[i]/total2*100 >= 2:
        new_labels2.append(labels[i])
        new_values2.append(values[i])
    else:
        etc+=values[i]

new_labels2.append('기타')
new_values2.append(etc)

wedgeprops = {'width':0.6}
explode = [0.1]*len(new_values2)
colors = ['dimgrey','grey','darkgrey','silver','lightgray','whitesmoke','white']
plt.pie(new_values2,labels = new_labels2, autopct = '%.1f%%',colors = colors,textprops = {'fontsize':8},startangle = 90,explode = explode,wedgeprops = wedgeprops)
plt.title('아파트에서 일어나는 범죄',color = 'red', fontsize = 7)
plt.show()
```

```python
residential = [
    '단독주택', '아파트', '다세대주택', '오피스텔'
]

commercial = [
    '백화점', '대형할인점', '슈퍼마켓', '편의점',
    '노점상', '무인상점', '기타상점',
    '음식점', '카페', '주점', '유흥업소',
    '숙박업소', '목욕시설', '미용실', '마사지업소'
]

transport = [
    '고속도로', '자동차 전용도로', '일반도로', '길거리',
    '버스정류소', '지하철역', '기차역', '여객선터미널', '공항',
    '버스', '택시', '자가용자동차', '지하철내부', '기차', '선박', '비행기'
]
```

```python
df_delete['주거공간'] = df[residential].sum(axis=1)
df_delete['상업공간'] = df[commercial].sum(axis=1)
df_delete['교통공간'] = df[transport].sum(axis=1)
df_delete
```

```python
df_delete.reset_index()
```

```python
df_delete = df_delete.groupby('범죄대분류').sum()
```

```python
import numpy as np
plt.figure(figsize = (15,5))
w = 0.25
x = np.arange(len(df_delete.index))
plt.title('공간 유형별 발생 범죄 수',color = 'red')
plt.bar(x-w,df_delete['주거공간'],width = 0.25,label = '주거공간에서 발생하는 범죄의 수',color = 'darksalmon')
plt.bar(x,df_delete['상업공간'],width = 0.25,label = '상업공간에서 발생하는 범죄의 수',color = 'coral')
plt.bar(x+w,df_delete['교통공간'],width = 0.25,label = '교통공간에서 발생하는 범죄의 수',color = 'lightsalmon')
plt.xticks(x,df_delete.index,ha = 'center',fontsize = 10)
plt.legend(fontsize = 15)
plt.grid(ls = ':')
plt.show()
```
